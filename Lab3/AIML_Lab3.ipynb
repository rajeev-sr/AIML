```python
#to convert mp3 to wav
from pydub import AudioSegment
import os

# Define the input MP3 file and output WAV file paths
input_mp3_file = "/content/my_noisy_audio.mp3"
output_wav_file = "my_noisy_audio.wav"

# Check if the input file exists
if not os.path.exists(input_mp3_file):
    print(f"Error: Input file not found at {input_mp3_file}")
else:
    # Load the MP3 file and export it as WAV
    try:
        audio = AudioSegment.from_file(input_mp3_file, format="mp3")
        audio.export(output_wav_file, format="wav")
        print(f"Successfully converted {input_mp3_file} to {output_wav_file}")
    except Exception as e:
        print(f"Error during conversion: {e}")
```

```python
import noisereduce as nr
import soundfile as sf
import numpy as np

input_file = "my_noisy_audio.wav"
output_file = "my_cleaned_audio.wav"

# Read audio data
data, rate = sf.read(input_file)

# If stereo, convert to mono by averaging channels
if len(data.shape) > 1:
    data = np.mean(data, axis=1)

# Here we use the first 0.5s as noise reference
reduced_noise = nr.reduce_noise(y=data, sr=rate, prop_decrease=1.0)

sf.write(output_file, reduced_noise, rate)

print(f"Noise-reduced audio saved as: {output_file}")
```

```python
import speech_recognition as sr

# Initialize the recognizer
r = sr.Recognizer()
```

```python
def aud_to_text(filename):
  # Transcribe the audio

  with sr.AudioFile(filename) as source:
    # listen for the data (load audio to memory)
    audio_data = r.record(source)

    # recognize (convert from speech to text)
    try:
        MyText = r.recognize_google(audio_data)
        MyText = MyText.lower()
        return MyText
    except sr.RequestError as e:
        print("Could not request results; {0}".format(e))
    except sr.UnknownValueError:
        print("Unknown error occurred")
```

```python
noisy_file = "/content/my_noisy_audio.wav"
noisy_cleaned_file = "/content/my_cleaned_audio.wav"
clean_file = "/content/my_clean_audio.wav"
```

```python
noisy_text=aud_to_text(noisy_file)
noisy_cleaned_text=aud_to_text(noisy_cleaned_file)
clean_text=aud_to_text(clean_file)
```

```python
print("Noisy Transcription:\n", noisy_text)
print("\nNoisy Cleaned Transcription:\n", noisy_cleaned_text)
print("\nClean Transcription:\n", clean_text)
```

```python
import jiwer

# Define the ground truth (clean transcription)
ground_truth = clean_text

# Calculate WER for noisy transcription
noisy_wer = jiwer.wer(ground_truth, noisy_text)
print(f"Word Error Rate (WER) for Noisy Transcription: {noisy_wer:.2f}\n")

# Calculate WER for cleaned transcription
cleaned_wer = jiwer.wer(ground_truth, noisy_cleaned_text)
print(f"Word Error Rate (WER) for Cleaned Transcription: {cleaned_wer:.2f}")
```
